```python
# gradient descent in linear regression
import numpy as np
```

```python
x=np.array([0.8,1,1.2,1.4,1.6,1.8,2,2.2,2.4,2.6])
y_hat=np.array([0.7,0.65,0.9,0.95,1.1,1.15,1.2,1.4,1.55,1.5])
```

```python
# initializing learning rate and parameters
b0 = 0.35      
b1 = 0.65

cost_hist = []
lr = 0.01
m = len(x)
epochs = 100

for i in range(epochs):
    y = b0 + b1*x
    cost = (1/2*m)*sum(y - y_hat)**2 # if you do y_hat - y then you have to add minus at the start
    db0 = (1/m)*sum(y - y_hat)
    db1 = (1/m)*sum((y - y_hat)*x)

    if i % 10 == 0:  # shows every 10th iteration
        print(f"Cost:{cost} \t interation: {i}")
    b0 = b0 - lr*db0         # changes weight value after every iteration otherwise it would be the same
    b1 = b1 - lr*db1
    cost_hist.append(cost)
print()
print(b0,b1)
```

```python
# plot the error
import matplotlib.pyplot as plt
cost_hist1 = np.asarray(cost_hist)
plt.plot(cost_hist1)
```

```python
def LGD(b0,b1,lr,x,y,m,epochs):
    b0 = b0
    b1 = b1

    cost_hist = []
    lr = lr

    epochs = epochs

    for i in range(epochs):
        y = b0 + b1*x
        cost = (1/2*m)*sum(y - y_hat)**2 # if you do y_hat - y then you have to add minus at the start
        db0 = (1/m)*sum(y - y_hat)
        db1 = (1/m)*sum((y - y_hat)*x)

        if i % 10 == 0:  # shows every 10th iteration
            print(f"Cost:{cost} \t interation: {i}")
        b0 = b0 - lr*db0         # changes weight value after every iteration otherwise it would be the same
        b1 = b1 - lr*db1
        cost_hist.append(cost)
    print()
    print(b0,b1)
```

```python
LGD(b0,b1,lr,x,y_hat,m,epochs)
```

```python
LGD(0.35,0.65,0.01,x,y_hat,m,100)
```

```python
# cross verify with sklearn libarary
import sklearn.linear_model as lm
lr = lm.LinearRegression()
x = x.reshape(-1,1)
lr.fit(x,y_hat)
print(lr.intercept_)
print(lr.coef_)
```

```python
# plot the fit line
plt.scatter(x,y_hat,label = "Data")
plt.plot(x, b1*x + b0, color = "red", label = "Model Prediction")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Descent")
plt.legend()
plt.grid()
plt.show()
```

```python
# # Lab 2 find out parameter theta0 theta1 and theta2 through gradient descent
import numpy as np
from sklearn.linear_model import SGDRegressor
x1=np.array([0.4,0.8,1,1.2,1.8,1.8,2,2,2.4,2])
x2=np.array([0.8,1,1.2,1.4,1.6,1.8,2,2.2,2.4,2.6])
y=np.array([0.7,0.65,0.9,0.95,1.1,1.15,1.2,1.4,1.55,1.5])
X = np.column_stack((x1, x2))
 
sgd = SGDRegressor(
    loss="squared_error",
    learning_rate="constant",
    eta0=0.01,
    max_iter=5000,
    tol=1e-6,
    random_state=42
)
 
sgd.fit(X, y)
 
print("Intercept:", sgd.intercept_[0])
print("w1 =", sgd.coef_[0])
print("w2 =", sgd.coef_[1])
```
