```python
# Linear regression gradient descent through matrix
import numpy as np
np.random.seed(5)
x = np.array([[5,3],[6,4],[7,6],[9,8]])
y = np.array([2,4,7,8]).reshape(-1,1)
x.shape # sample no, features
y.shape
```

```python
def LGR(x,y,lr,epoch):
    nx = x.shape[1]  # Number of features
    m = x.shape[0]  # Number of traning samples
    b = np.random.rand(nx,1) # Random initailization for weights
    b0 = 0  # initial bias
    cost = []  # STores ow value changes, should decrease
    for i in range(epoch):
        y_hat = np.dot(x,b) + b0
        error = y_hat - y
        cost1 = np.dot(x.T,error)/(2*m)
        cost.append(cost1)
        dw = np.dot(x.T,error)/m
        db = np.sum(y_hat - y)/m
        b = b - lr*dw
        b0 = b0 - lr*db
    return b0,b
```

```python
np.random.seed(5)
x = np.array([[5,3],[6,4],[7,6],[9,8]])
y = np.array([2,4,7,8]).reshape(-1,1)
LGR(x,y,0.01,10000)
```

```python
import numpy as np
import matplotlib.pyplot as plt
nx = x.shape[1]  # Number of features
m = x.shape[0]  # Number of traning samples
b = np.random.rand(nx,1) # Random initailization for weights
b0 = 0  # initial bias
lr = 0.01
cost = []  # STores ow value changes, should decrease
for i in range(10000):
    y_hat = np.dot(x, b) + b0
    error = y_hat - y
    cost1 = np.sum(error**2) / (2 * m)
    cost.append(cost1)
    dw = np.dot(x.T, error) / m
    db = np.sum(error) / m
    if i % 100 == 0:  # shows every 10th iteration
        print(f"Cost: {cost1:.6f} \t iteration: {i}")
    b = b - lr * dw
    b0 = b0 - lr * db
print(b, b0)
cost_hist1 = np.asarray(cost)
plt.plot(cost_hist1)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")
plt.show()
```
