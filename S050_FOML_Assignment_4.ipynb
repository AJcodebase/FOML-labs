```python
def sigmoid(z):
    s = 1/(1 + np.exp(-z))
    return s
```

```python
x = np.array([[1.,2.,-1.],[3.,4.,-3.2]])
y = np.array([[1,0,1]])
x = x.T
y = y.T
w = np.array([[1.2],[2.3]])
print(w.shape)
b = 2
nx = 2
m = 3
h = np.dot(x,w) + b
a = sigmoid(h)
error = a-y
cost = -(np.sum(y * np.log(a) + (1-y)* (np.log(1-a))))/m
dw = np.dot(x.T, error) / m
db = np.sum(a-y) / m
lr = 0.09
w = w - lr * dw
b = b - lr * db
print(w,b,h)
```

```python
def LGD(x,y,lr,epoch):
    x = x
    y = y
    x = x.T
    y = y.T
    b = 0
    nx = x.shape[1]  # Number of features
    m = x.shape[0]  # Number of traning samples
    w = np.random.rand(nx,1)
    cost = []  # STores ow value changes, should decrease
    for i in range(epoch):
        h = np.dot(x,w) + b
        a = sigmoid(h)
        a = np.clip(a,1e-15,1-1e-15)
        error = a-y
        cost1 = -(np.sum(y * np.log(a) + (1-y)* (np.log(1-a))))/m
        if i % 1000 == 0:
            print(cost1)
        cost.append(cost1)
        dw = np.dot(x.T, error) / m
        db = np.sum(error) / m
        w = w - lr * dw
        b = b - lr * db
    return(w,b)
```

```python
x = np.array([[1.,2.,-1.],[3.,4.,-3.2]])
y = np.array([[1,0,1]])
LGD(x,y,0.01,10000)
```
