```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(1)
x = np.random.randn(2,3)
print(x)
```

```python
def sigmoid(z):
    s =1/1+np.exp(-z)
    return s
```

```python
s = sigmoid(x)
print(s)
```

```python
def tan(z):
    s = (np.exp(z)-np.exp(-z))/(np.exp(z)+np.exp(-z))
    return s
```

```python
t = tan(x)
print(t)
```

```python
def layer_sizes(x, y,layers):
    n_x = x.shape[0]
    n_h = layers
    n_y = y.shape[0]
    return (n_x, n_h, n_y)
```

```python
n_x, n_h, n_y = layer_sizes(x,y,4)
```
