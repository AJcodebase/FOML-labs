```python
def layer_sizes(x, y,layers):
    n_x = x.shape[0]
    n_h = layers
    n_y = y.shape[0]
    return (n_x, n_h, n_y)
```

```python
n_x, n_h, n_y = layer_sizes(x,y,4)
```

```python
print(n_x, n_h, n_y)
```

```python
def initialize_parameters(n_x, n_h, n_y):
    np.random.seed(2)
    W1 = np.random.randn(n_h,n_x)*0.01
    b1 = np.zeros((n_h,1))
    W2 = np.random.randn(n_y,n_h)*0.01
    b2 = np.zeros((n_y,1))
    
    parameters = {"W1": W1,
                  "b1": b1,
                  "W2": W2,
                  "b2": b2}
    
    return parameters
```

```python
p = initialize_parameters(n_x, n_h, n_y)
```

```python
p
```

```python
def forward_propogation(x,p):
    W1 = p["W1"]
    b1 = p["b1"]
    W2 = p["W2"]
    b2 = p["b2"]
    
    Z1 = np.dot(W1,x)+b1
    A1 = tan(Z1)
    Z2 = np.dot(W2,A1)+b2
    A2 = sigmoid(Z2)
    
    cache = {"Z1": Z1,
             "A1": A1,
             "Z2": Z2,
             "A2": A2}
    
    return A2, cache
```

```python
A2, cache = forward_propogation(x,p)
```

```python
A2
```

```python
cache
```

```python
def compute_cost(A2,y,p):
    m = y.shape[1]
    logprobs = np.multiply(np.log(A2),y)+np.multiply(np.log(1-A2),(1-y))
    cost = -np.sum(logprobs)/m
    
    cost = float(np.squeeze(cost))
    
    return cost
```

```python
compute_cost(A2,y,p)
```
